```python
import os
import glob
import pandas as pd

file_list = glob.glob('Country/*.xlsx')

df_list = []
for file in file_list:
    df = pd.read_excel(file)

    df['country'] = os.path.splitext(os.path.basename(file))[0]

    df_list.append(df)

df_combined = pd.concat(df_list, ignore_index=True)

df_combined.head()
```

```python
df_combined = df_combined.iloc[1:]
df_combined.head()
```

```python
df_combined.columns = df_combined.iloc[0]

df_combined = df_combined[1:]

df_combined.head()
```

```python
df_combined = df_combined.drop(columns=['2024-2025', '%Growth', 'Commodity', 'S.No.'])
```

```python
df_combined = df_combined.rename(columns={'angola': 'country', 'HSCode': 'hs4_code', '2025-2026': 'import_sales'})
```

```python
df_combined.head()
```

```python
df_combined['hs4_code'] = df_combined['hs4_code'].astype(str)

df_combined['hs2_code'] = df_combined['hs4_code'].str[:2]
df_combined.head()
```

```python
df_commodity = pd.read_excel('dim_commodity.xlsx')
df_commodity.head()
```

```python
df_combined['hs4_code'] = df_combined['hs4_code'].astype(str).str.strip()
df_commodity['hs4_code'] = df_commodity['hs4_code'].astype(str).str.strip().str.zfill(4)

merged_df = pd.merge(df_combined, df_commodity, on='hs4_code')
merged_df.head()
```

```python
merged_df.head(40)
```

```python
merged_df.loc[merged_df['country'] == 'usa', 'country'] = 'U S A'
merged_df.loc[merged_df['country'] == 'china', 'country'] = 'CHINA P RP'
merged_df.loc[merged_df['country'] == 'uae', 'country'] = 'U ARAB EMTS'
merged_df.loc[merged_df['country'] == 'uk', 'country'] = 'U K'

merged_df['country'] = merged_df['country'].str.upper()
```

```python
df_country = pd.read_excel('dim_country.xlsx')
df_country.head()
```

```python
final_df = pd.merge(merged_df, df_country, on='country')
final_df.head()
```

```python
final_df = final_df[final_df['import_sales'] != 0]
```

```python
final_df = final_df.drop(columns=['hs4_code', 'country', 'hs2_code_x', 'hs2_code_y', 'commodity', 'description', '2025_2026_import_sales_usd', 'region', 'continent', 'import_sales_usd_country_2025_2026'])
```

```python
final_df.to_excel('fact_imports.xlsx', index=False)
```
