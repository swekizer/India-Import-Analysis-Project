```python
#Importing the excel file
import pandas as pd

df = pd.read_excel('TradeStat-Eidb-Import-Commodity-wise.xlsx', dtype=str)

df.head()
```

```python
#Deleting the first row
df = df.iloc[1:]
df.head()
```

```python
#Moving first row to header. Fixing the attributes
df.columns = df.iloc[0]
df = df[1:]
df.reset_index(drop=True, inplace=True)
```

```python
df.head()
```

```python
#Removing unwanted colums 2024-2025, %Share, %Growth
df = df.drop(columns=["2024 - 2025", "%Share", "%Growth"])
df.head()
```

```python
#Rename the columns
df = df.rename(columns={'S.No.': 'commodity_key', 'HSCode': 'hs4_code', 'Commodity': 'description', '2025 - 2026': '2025_2026'})
df.head()
```

```python
#Creating a new columns of just first two numbers in hs4_code
df["hs2_code"] = df["hs4_code"].str[:2]
df.head()
```

```python
df_hs2 = pd.read_excel('TradeStat-Eidb-Import-Commodity-wise-hs2.xlsx', dtype=str)
df_hs2.head()
```

```python
#Deleting the first row
df_hs2 = df_hs2.iloc[1:]
df_hs2.head()
```

```python
#Moving first row to header. Fixing the attributes
df_hs2.columns = df_hs2.iloc[0]
df_hs2 = df_hs2[1:]
df_hs2.reset_index(drop=True, inplace=True)
df_hs2.head()
```

```python
#Removing unwanted colums 2024-2025, %Share, %Growth
df_hs2 = df_hs2.drop(columns=["2024 - 2025", "%Share", "%Growth", "2025 - 2026"])
df_hs2.head()
```

```python
#Rename the columns
df_hs2 = df_hs2.rename(columns={'S.No.': 'chapter_key', 'HSCode': 'hs2_code', 'Commodity': 'commodity'})
df_hs2.head()
```

```python
merged_df = pd.merge(df,df_hs2, on="hs2_code")
merged_df.head()
```

```python
#Removing unwanted colums
merged_df = merged_df.drop(columns=["chapter_key"])
merged_df.head()
```

```python
#Rearranging the columns
merged_df = merged_df[["commodity_key", "hs2_code", "hs4_code", "commodity", "description", "2025_2026"]]
merged_df.head()
```

```python
#Rename the columns
merged_df = merged_df.rename(columns={'2025_2026': '2025_2026_import_sales_usd'})
merged_df.head()
```

```python
merged_df['2025_2026_import_sales_usd'] = pd.to_numeric(merged_df['2025_2026_import_sales_usd'])
```

```python
merged_df.to_excel('dim_commodity.xlsx', index=False)
```
